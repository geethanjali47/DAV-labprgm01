```python
import pandas as pd
data = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena', 'Arjun'],
    'Age': [21, 22, None, 23, 21, None],
    'Salary': [25000, None, 30000, 28000, None, 32000],
    'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', 'ECE'],
    'City': ['Mysore', 'Bangalore', 'Mysore', 'Hubli', 'Bangalore', 'Mysore']
}
df = pd.DataFrame(data)
print("Original Data:")
print(df)
median_salary = df['Salary'].median()
print("\nMedian Salary:", median_salary)
df['Salary'] = df['Salary'].fillna(median_salary)
print("\nData after Median Imputation:")
print(df)
```

```python
import pandas as pd
data = {
    'Age': [20, 25, None, 30, 35]
}
df = pd.DataFrame(data)
print("Original Data:")
print(df)
mean_age = df['Age'].mean()
df['Age'] = df['Age'].fillna(mean_age)
print("\nData after Mean Imputation:")
print(df)
print("\nMean Age:", mean_age)

```
